```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split,GridSearchCV,cross_val_score
from sklearn.tree import DecisionTreeRegressor,DecisionTreeClassifier
from sklearn.linear_model import LinearRegression,LogisticRegression
from sklearn.ensemble import BaggingClassifier,BaggingRegressor
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
from sklearn.metrics import r2_score,accuracy_score
from sklearn.datasets import load_iris,make_classification,fetch_california_housing
from sklearn.svm import SVC,SVR
```

```python
X,y=make_classification(n_samples=1000,n_features=10,n_informative=3)
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
dt=DecisionTreeClassifier()
dt.fit(x_train,y_train)
accuracy_score(y_test,dt.predict(x_test))
```

```python
bag=BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=100,
    max_samples=0.25,
    bootstrap=False,
    random_state=42
)
```

```python
bag.fit(x_train,y_train)
accuracy_score(y_test,bag.predict(x_test))
```

```python
svc=SVC()
svc.fit(x_train,y_train)
accuracy_score(y_test,svc.predict(x_test))
```

```python
bag=BaggingClassifier(
    estimator=SVC(),
    n_estimators=500,
    max_samples=0.25,
    bootstrap=True,
    random_state=42
)
```

```python
bag.fit(x_train,y_train)
accuracy_score(y_test,bag.predict(x_test))
```

```python
bag=BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=500,
    max_samples=1.0,
    bootstrap=True,
    random_state=42,
    max_features=0.5,
    oob_score=True            ###oob score-- out of bag score, helps because if boostrap=True, sometimes a particualr row might not be in any row or some might be in all, statistically only 63% are used. So oob parameter utilises the rest of 37%
)
```

```python
cross_val_score(bag,X,y,cv=5).mean()
```
