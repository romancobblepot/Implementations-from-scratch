```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,LogisticRegression
from sklearn.metrics import r2_score,mean_squared_error
```

```python
from sklearn.datasets import make_classification
X,y = make_classification(n_samples=100, n_features=2,n_informative=1,n_redundant=0, n_classes=2, n_clusters_per_class=1,random_state=41)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
def step(z):
  return 1/(1+np.exp(-z))
```

```python
def perceptron(X,y):
  X=np.insert(X,0,1,axis=1)
  w=np.ones(X.shape[1])
  lr=0.1
  for i in range(10000):
    x_random=np.random.randint(0,X.shape[0])
    x=X[x_random]
    y_pred=step(np.dot(x,w))
    w+=(lr*(y[x_random]-y_pred)*x)
  return w

```

```python
perceptron(X,y)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=41)
```

```python
lrc=LogisticRegression()
lrc.fit(X_train,y_train)
```
