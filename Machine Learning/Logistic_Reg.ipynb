```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,LogisticRegression
from sklearn.metrics import r2_score,mean_squared_error,precision_score,recall_score,f1_score,accuracy_score
```

```python
from sklearn.datasets import make_classification
size=3
raw_array = np.random.rand(size)            ###Standard_Normalization
random_weights = raw_array / np.sum(raw_array)
X,y = make_classification(n_samples=100, n_features=2,n_informative=2,n_redundant=0,weights=random_weights,n_classes=size, n_clusters_per_class=1,random_state=43)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=41)
df=pd.Series(y_train)
df.value_counts()
```

```python
lrc=LogisticRegression()
lrc.fit(X_train,y_train)
lrc.predict(X_test)
```

```python
lrc.coef_
```

```python
lrc.intercept_
```

```python
print(precision_score(y_test,lrc.predict(X_test),average='weighted'))
print(recall_score(y_test,lrc.predict(X_test),average='weighted'))
print(f1_score(y_test,lrc.predict(X_test),average='weighted'))
print(accuracy_score(y_test,lrc.predict(X_test)))
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,lrc.predict(X_test)))
```

```python
class gradient_descent_logistic():
  def __init__(self,learning_rate,epochs):
    self.learning_rate=learning_rate
    self.epochs=epochs
    self.coef_=None
    self.intercept_=None
  def sigmoid(self,z):
    return 1/(1+np.exp(-z))
  def fit(self,X,y):
    X=np.insert(X,0,1,axis=1)
    self.coef_=np.ones(X.shape[1])
    for i in range(self.epochs):
      y_pred=self.sigmoid(np.dot(X,self.coef_))
      coef_der=np.dot(y_pred-y,X)/X.shape[0]
      self.coef_-=self.learning_rate*(coef_der)
    self.intercept_=self.coef_[0]
    self.coef_=self.coef_[1:]
  def predict(self,X,threshold):
    y_pred=self.sigmoid(np.dot(X,self.coef_)+self.intercept_)
    y_pred=np.where(y_pred>threshold,1,0)
    return y_pred




```

```python
gdlrc=gradient_descent_logistic(learning_rate=0.1,epochs=250)
gdlrc.fit(X_train,y_train)
gdlrc.predict(X_test,0.5)
```
