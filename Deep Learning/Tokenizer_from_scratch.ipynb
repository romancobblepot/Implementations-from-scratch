```python
with open("/content/the-verdict.txt","r",encoding="utf-8") as f:
  input=f.read()
```

```python
import re
```

```python
result=re.split(r'(--|[.,:?_!"()\']|\s)',input)
preprocessed_input=[item.strip() for item in result if item.strip()]
preprocessed_input.append("<unk>")
preprocessed_input.append("<|endoftext|>")
preprocessed_input.append("<|pad|>")
preprocessed_input.append("<sos>")
```

```python
vocab={token:integer for integer,token in enumerate(set(preprocessed_input))}
```

```python
class Tokenizer():
  def __init__(self,vocab):
    self.vocab=vocab
    self.id_to_token={id:token for token,id in vocab.items()}
  def encode(self,x):
    result=re.split(r'(--|[.,:?_!"()\']|\s)',x)
    preprocessed_x=[item for item in result if item and item.strip()]
    encoded_ids=[]
    for item in preprocessed_x:
      if item not in self.vocab:
        encoded_ids.append(self.vocab["<unk>"])
      else:
        encoded_ids.append(self.vocab[item])
    return encoded_ids,preprocessed_x
  def decode(self,token_ids):
    text=" ".join([self.id_to_token[id] for id in token_ids])
    result=re.sub(r'\s+([,.?!"()\'])',r'\1',text)
    return result


```

```python
instance=Tokenizer(vocab)
encoded_result,preprocessed_x=instance.encode("I am baljeet")
```
